```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
df.to_csv('Iris[1].csv', index=False)
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.patches as mpatches

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Visualization 1: Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
print(f"Accuracy: {accuracy_score(y_test, y_pred)*100:.1f}%")

# Visualization 2: Workflow Diagram
fig, ax = plt.subplots(figsize=(8, 3))
ax.axis('off')
steps = ["Data\n(Iris)", "Train\n(80%)", "Model", "Predict", "Accuracy"]
for i, s in enumerate(steps):
    box = mpatches.FancyBboxPatch((i*1.6, 0.5), 1.3, 0.8, boxstyle="round,pad=0.1", fc="lightblue", ec="navy")
    ax.add_patch(box)
    ax.text(i*1.6+0.65, 0.9, s, ha='center', va='center', fontsize=9, fontweight='bold')
    if i < len(steps)-1:
        ax.arrow(i*1.6+1.3, 0.9, 0.25, 0, head_width=0.08, fc='black')
plt.title('ML Workflow Diagram')
plt.xlim(-0.2, 8)
plt.ylim(0, 2)
plt.show()
```
