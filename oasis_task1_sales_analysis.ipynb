```python
df = pd.read_csv('retail_sales_dataset.csv')
print(df.head())
print(df.shape)
```

```python
# Missing values entha unnayo chuddam
print(df.isnull().sum())

# Customer satisfaction and discount ni fill cheddam
df['discount_pct'].fillna(0, inplace=True)
df['customer_satisfaction'].fillna(df['customer_satisfaction'].median(), inplace=True)
print("Cleaning ayyindi!")
```

```python
import matplotlib.pyplot as plt

# Product category wise sales chuddam
df.groupby('product_category')['sales_amount'].sum().plot(kind='bar')
plt.title('Category wise Sales')
plt.show()

print("Graph vachinda?")
```

```python
df.groupby('product_category')['sales_amount'].sum().plot(kind='bar')
plt.title('Category wise Sales')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```
